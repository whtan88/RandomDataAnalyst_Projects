```python
# Final Code that combines all the above codes into a playable minesweeper game

# start with creating the board size
def minesweeper(difficulty):
  if difficulty == 'E':
    size = (8, 8)
    mines = 10
  elif difficulty == 'M':
    size = (16, 16)
    mines = 40
  elif difficulty == 'H':
    size = (30, 16)
    mines = 99
  return size, mines

# start with creating the board size
print('Enter difficulty of minesweeper game (E, M or H)')
difficulty = input().upper()

# initialize and create the base board
size, mines = minesweeper(difficulty)
board_ui = [['_'] * size[1] for _ in range(size[0])]

# create the backend board with mines populated
board_backend = [['_'] * size[1] for _ in range(size[0])]
mines_location_list = []

# randomize the location of the mine, ensuring that there is no duplicate coordinates and all the mines are planted on different coordinates
for _ in range(mines):
  mines_location = (random.randint(0, size[0] - 1), random.randint(0, size[1] - 1))
  while mines_location in mines_location_list:
    mines_location = (random.randint(0, size[0] - 1), random.randint(0, size[1] - 1))
  board_backend[mines_location[0]][mines_location[1]] = 'X'
  mines_location_list.append(mines_location)

# compute the rest of the field values for the backend board
for i in range(size[0]):
  for j in range(size[1]):
    mine_counter = 0
    if board_backend[i][j] != 'X':
      for x in range(i - 1, i + 2):
        for y in range(j - 1, j + 2):
          if 0 <= x < size[0] and 0 <= y < size[1] and board_backend[x][y] == 'X':
            mine_counter += 1
      board_backend[i][j] = str(mine_counter)

# record trackers
move_counter = 0
game_status = 1
no_mines_counter = mines

while game_status == 1:
  start_time = time.perf_counter()
  if move_counter == 0:
    game_start_time = start_time

  # updated display board with axis and print results with numbr of moves, no. of mines and elapsed time
  list_y = [str(i + 1) for i in range(size[0])]
  list_x = [str(i + 1) for i in range(size[1])]

  print(f"No. of moves : {move_counter}   No. of mines : {no_mines_counter}   Elapsed Time : {round(time.perf_counter() - game_start_time,2)} sec")
  print(f"{'-' * 50}")
  print(f"ax{list_x}")
  c = 0
  for i in board_ui:
    print(f"{list_y[c]} {i}")
    c += 1
  print(f"{'-' * 50}")

  print('Enter move here in the following format : F,x,y or D,x,y (where F is a flag or D is to dig, with x and y being the coordinates)')
  move_input = input().upper().split(',')

  if len(move_input) != 3:
      print("Invalid input format. Please use F,x,y or D,x,y")
      continue

  action = move_input[0]
  try:
    # Adjusting for 0-based indexing
    choice_cell = [int(move_input[2]) - 1, int(move_input[1]) - 1]
  except ValueError:
      print("Invalid coordinates. Please enter integers for x and y.")
      continue

  if not (0 <= choice_cell[0] < size[0] and 0 <= choice_cell[1] < size[1]):
      print("Coordinates out of bounds.")
      continue

  end_time = time.perf_counter()
  elapsed_time = end_time - game_start_time
  move_counter += 1

  # determine action to be taken on the board, plant a flag or dig the cell
  if action == 'F':
    if board_ui[choice_cell[0]][choice_cell[1]] == '_':
      board_ui[choice_cell[0]][choice_cell[1]] = 'F'
      no_mines_counter -= 1
    elif board_ui[choice_cell[0]][choice_cell[1]] == 'F':
      board_ui[choice_cell[0]][choice_cell[1]] = '_'
      no_mines_counter += 1
  elif action == 'D':
    # lose condition
    if board_backend[choice_cell[0]][choice_cell[1]] == 'X':
      print('Game over, you lost.')
      board_ui = board_backend
      game_status = 0

    else:
      # create a list of coordinates of adjacent cells and the cell the player selected
      list_to_reveal_tiles = []
      for i in [-1,0,1]:
        for j in [-1,0,1]:
          cell = [choice_cell[0] + i, choice_cell[1] + j]
          if 0 <= cell[0] < size[0] and 0 <= cell[1] < size[1]:
            list_to_reveal_tiles.append(cell)

      # randomly select a number of cells to reveal
      lower = max(3, math.ceil(0.33 * len(list_to_reveal_tiles)))
      upper = math.ceil(0.67 * len(list_to_reveal_tiles))

      # If lower is greater than upper, set no_revealed_tiles to lower
      if lower > upper:
          no_revealed_tiles = lower
      else:
          no_revealed_tiles = random.randint(lower, upper)

      # revealing the chosen cell
      if board_ui[choice_cell[0]][choice_cell[1]] == '_':
          board_ui[choice_cell[0]][choice_cell[1]] = board_backend[choice_cell[0]][choice_cell[1]]
          no_revealed_tiles -= 1
          if choice_cell in list_to_reveal_tiles:
              list_to_reveal_tiles.remove(choice_cell)

      # revealing a number of randomly selected cells
      while no_revealed_tiles > 0 and list_to_reveal_tiles:
        cell_to_reveal = random.choice(list_to_reveal_tiles)
        if board_backend[cell_to_reveal[0]][cell_to_reveal[1]] != 'X':
          board_ui[cell_to_reveal[0]][cell_to_reveal[1]] = board_backend[cell_to_reveal[0]][cell_to_reveal[1]]
          list_to_reveal_tiles.remove(cell_to_reveal)
          no_revealed_tiles -= 1
        elif board_backend[cell_to_reveal[0]][cell_to_reveal[1]] == 'X':
          list_to_reveal_tiles.remove(cell_to_reveal)


      # check win condition
      unrevealed_cells = 0
      for r in range(size[0]):
          for c in range(size[1]):
              if board_ui[r][c] == '_' or board_ui[r][c] == 'F':
                  unrevealed_cells += 1

      if unrevealed_cells == mines:
          print('Congratulations, you won!')
          # Reveal all mines as flags
          board_ui = board_backend
          for r, c in mines_location_list:
              board_ui[r][c] = 'F'
          game_status = 0

print(f"No. of moves : {move_counter}   No. of mines : {no_mines_counter}   Elapsed Time : {round(time.perf_counter() - game_start_time,2)} sec")
print(f"{'-' * 50}")
print(f"ax{list_x}")
c = 0
for i in board_ui:
  print(f"{list_y[c]} {i}")
  c += 1
print(f"{'-' * 50}")
```
